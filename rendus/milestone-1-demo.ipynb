## Objectifs pédagogiques

- Rechercher en autonomie de l'information sur une notion et le code correspondant
- Rédiger un notebook synthétisant les éléments clés
- Transmettre les idées principales à l'oral

## Consignes

Chaque groupe **remet un notebook** démonstratif et le **présente** oralement en cours (10 minutes).  
La démo porte sur la thématique assignée et doit la rendre accessible et compréhensible pour l'ensemble de la classe : l'objectif n'est pas de tout couvrir en profondeur, mais de transmettre clairement l'essentiel à des pairs qui découvrent la notion.

Structure suggérée (What / Why / How / Resources) :

- What : de quoi s'agit-il ? (définition, principe général de la méthode ou de l'outil)
- Why : à quoi ça sert ? dans quels cas d'usage est-ce pertinent ?
- How : comment l'appliquer ? démonstration pas à pas sur un jeu de données, avec du code commenté et exécutable
- Resources : pour aller plus loin (documentation, tutoriels, lectures complémentaires) + références utilisées pour la construction du contenu.
