# Fiche de lecture

## Nom & Prénom

- Étudiant.e #1

## Objectifs pédagogiques

- Analyser et synthétiser un texte scientifique
- Maîtriser ce qu’est une problématique la reconnaître dans un texte
- Maîtriser les concepts nécessaires à la création de la problématique de recherche personnelle
- Comprendre le schéma d’argumentation d’un texte scientifique
- Identifier les données utilisées
- Identifier les méthodes computationnelles mobilisées et les enjeux afférents


## Consignes

- Identifier 1 article (ou chapitre) de référence utilisant les sciences sociales computationnelles.
- Celui-ci peut être en lien avec le sujet de mémoire envisagé, ou plus générique.
- Réaliser une fiche synthétique de lecture de deux pages permettant d'identifier :
  - la problématique
  - les données et les variables utilisées
  - les méthodes computationnelles mises en oeuvre
  - les enjeux soulevés par le recours à ces méthodes (intérêts, limites)
  - les principaux résultats en lien avec le projet de mémoire (si pertinent)

### Modalités

Rédaction individuelle d'un compte-rendu de lecture. Il s’agit de réaliser une revue critique d’un article scientifique ou d’un chapitre ouvrage. Il est utile de s’interroger sur les points suivants :

- Quel est le problème de départ (s’il y en a un) ?
- Quelles sont les références théoriques mobilisées ?
- Quels sont les concepts centraux mobilisés ? (= en donner la définition)
- Quelles sont les hypothèses (s’il y en a, explicites ou implicites) ?
- Quels sont les outils de recueil et d’analyse de données mobilisés ?
- Quelles méthodes sont mises en oeuvre ?
- Quels sont les données et les variables utilisées ?
- Quelle est la logique de l’argumentation ?
- Quelles sont les conclusions auxquelles parvien(nen)t le(s) auteur(s) ?
- Quel est l’apport de ce travail pour votre propre enquête ?

### Critères d’évaluation

- Rédaction d’un document propre dans le texte (grammaire/orthographe) et sur la forme (existence de parties, de paragraphes, etc.)
- Prise en compte des points clés : définitions, problématique, hypothèses (s’il y a en a), données, méthodes, argumentaire d’ensemble, etc.
- Rédaction d’un document synthétique (pas de blabla) et analytique (pas seulement résumer mais réfléchir à ce qu’on lit)
