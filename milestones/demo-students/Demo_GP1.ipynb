## <u> **Quoi** </u>

<u>Les structures de bases</u>  
Variables : sert à stocker une seule valeur en mémoire, on utilise le signe égal "="  
Listes : sert à stocker plusieurs valeurs ordonnées dans une seule variable, on utilise des crochets "[]"  
Dictionnaires : sert à stocker des données sous forme de paire clé-valeur, on utilise des accolades "{}"  


<u> Qu'est ce que Pandas ? </u>  
--> Une bibliothèque Open source de Python.
Elle sert à manipuler, analyser et traiter des données tabulaires.  
*Fun fact : le nom Pandas vient de la contraction du terme Panel Data*

Pandas repose sur deux structures de données principales : les Series et DataFrame.  
**Une Series** est une liste d'éléments avec un index, similaire à une colonne unique.  
**Un DataFrame** est une table à deux dimensions composée de plusieurs Series, similaire à un tableau.

## <u> **Pourquoi** </u>
Dans le cas des sciences sociales et plus particulièrement de l'analyse de données, Pandas permet d'importer des fichiers de données, d'accéder à des bases de données, pour ensuite pouvoir les analyser.

- Les structres de données permettent de passer de milliers d'informations dispersées à une structure organisée et exploitable.  
- Puis des pouvoirs les transformer et les analyser.  
- Et de rendre cette analyse reproductible en inscrivant les opérations dans un script.

**Objectif n°2** : apprendre à utiliser les listes.  

```python
# Pour déclarer une liste, il suffit de suivre le même procédé mais en utilisant des crochets et en séparant nos éléments par des virgules.
weekend = ["samedi","dimanche"]
print(weekend)
# Une liste est indéxée à partir de zéro, notre premier élément sera donc associé à 0, le second à 1, etc.
weekend[0]
```

**Cas de figure** : Je veux faire mes courses avec Python, rien de plus normal.

Je **crée donc ma liste** de course dans ce super notebook.


```python
# Je veux avoir ma liste de course sur python et je vais appeler ma liste fruits_légumes
fruits_légumes = ["pomme", "banane", "fraise", "patate", "tomate"]
print(fruits_légumes)
```

**Retrouver des éléments dans sa liste**.


```python
# Comment récupérer un élément précis de ma liste ? chaque élément porte un numéro/ un indice qui rpz sa position dans la liste
# Attention le premier numéro est 0, donc pomme = 0
print(fruits_légumes[0])

#Pour avoir le dernier élément de ma liste, je peux faire ceci
print(fruits_légumes[len(fruits_légumes) - 1])
```

**Modifier une variable**. Patates < patates douces !!!

```python
#Comment modifier une variable dans ma liste
#Je veux modifier la variable 'patate' pour qu'elle devienne 'patate douce'
fruits_légumes [3] = "patate douce"
print(fruits_légumes)
```

**Rajouter une variable**. Oupss j'ai oublié de noter de prendre des pêches..


```python
#Je veux maintenant insérer une variable dans ma liste
#Je veux mettre 'pêche' en troisième dans ma liste
fruits_légumes.insert(2, "pêche")
print(fruits_légumes)
```

```python
#Je veux ajouter une variable à la fin de la liste
#Je veux penser à acheter des poivrons
fruits_légumes.append("poivrons")
print(fruits_légumes)
```

```python
#je veux ajouter plusieurs variables à la fin
#Je veux penser à acheter des abricots et des oranges
fruits_légumes.extend(["abricots","oranges"])
print(fruits_légumes)
```

**Supprimer une variable**.

```python
#Je veux supprimer une variable
#Je ne veux pas de pêche finalement
del fruits_légumes[2]
print (fruits_légumes)

#Je ne veux plus de banane
fruits_légumes.remove("banane")
print(fruits_légumes)
```

**Supprimer ma liste**. Course terminée, bravo à vous toustes !


```python
#j'ai terminé mes courses et je veux supprimer toute la liste
del fruits_légumes[2:3]
print(fruits_légumes)
```

**Objectif n°3** : Commment créer des dictionnaires?

```python
# Pour créer un dictionnaire, on peut partir de variables ou de listes, on nommera le dictionnaire puis on entrera les valeurs associées entre accolades.
#imaginons que je reprenne ma liste de course, mais que cette fois, je suis en Espagne parce que ça coûte quand même un peu moins cher, j'ai besoin de tarduire mes aliments
course = {
    "pomme": ["manzana"],
    "banane": ["banana"],
    "fraise":["frutilla"],
}

# Quand on crée un dictionnaire à partir de listes, on associe des clés (ici la langue), et des valeurs (ici les traductions)
nourriture = {
    "français": ["pomme", "banane", "fraise"],
    "espagnol": ["manzana", "banana", "frutilla"]
}

# Pour l'instant, si on affiche un dictionnaire, il ressortira sous une forme linéaire, et pas sous forme tabulaire.
nourriture
```

Comment importer Pandas

```python
# Par coutume et par mesure d'efficacité, en important la bibliothèque Pandas, on la définira par pd.
import pandas as pd
```

Comment construire des DataFrames

```python
# Utiliser Pandas va nous permettre d'avoir cet affichage sous forme tabulaire.
#Pour que ma liste de course soit un peu plus claire, et que je puisse faire mes achats en espagne sans soucis, j'aimerai avoir les traductions des aliments sous la forme de tableau

df_nourriture = pd.DataFrame(nourriture)
df_nourriture
```

```python
#je peux aussi décider de vouloir le dictionnaire non pas en fonction de la langue, mais en fonction des aliments que j'ai choisi

df_course = pd.DataFrame(course)
df_course
```

Je veux maintenant rajouter une ligne à mon tableau


```python
#je peux ajouter les oranges que j'avais oublié

df_nourriture.loc[len(df_nourriture)] = ["orange", "naranja"]
df_nourriture
```

Changer une variable

```python
# Je veux maintenant modifier la ligne des bananes, car j'ai appris qu'elles pouvaient aussi se traduire comme platano
df_nourriture.loc[1] = ["banane", "platano"]
df_nourriture
```

supprimer une ligne du dictionnaire

```python
#je veux enlever la ligne des fraises, car je les ai déjà achetées

df_nourriture.drop (2, inplace = True)
df_nourriture
```
